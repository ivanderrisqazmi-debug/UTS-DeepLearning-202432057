## BAB 2 — PEMAHAMAN ALGORITMA LSTM

### 2.1 Konsep kerja LSTM

Long Short-Term Memory (LSTM) merupakan pengembangan Recurrent Neural Network (RNN) yang dirancang untuk mempelajari hubungan dalam data berurutan. LSTM membawa dua informasi antarlangkah waktu, yaitu *hidden state* (`h_t`) dan *cell state* (`C_t`). Cell state berfungsi sebagai jalur memori, sedangkan beberapa gerbang mengatur informasi yang dilupakan, ditambahkan, dan dikeluarkan.

Pada proyek ini, teks tweet diubah menjadi urutan token berupa angka. Lapisan Embedding mengubah setiap token menjadi vektor, lalu LSTM memproses urutan tersebut. Hidden state terakhir diteruskan ke lapisan linear untuk menghasilkan tiga skor kelas: Negative, Neutral, dan Positive. Kelas dengan skor terbesar menjadi prediksi sentimen.

Empat komponen utama LSTM adalah:

1. **Forget gate** menentukan bagian memori sebelumnya yang dipertahankan.
2. **Input gate** menentukan seberapa banyak informasi baru yang dimasukkan.
3. **Candidate cell state** membentuk kandidat informasi baru.
4. **Output gate** menentukan informasi memori yang diteruskan sebagai hidden state.

Fungsi sigmoid `σ` menghasilkan nilai antara 0 dan 1, sedangkan `tanh` menghasilkan nilai antara -1 dan 1. Nilai gerbang mendekati 0 berarti informasi lebih banyak ditahan/dilupakan, sedangkan nilai mendekati 1 berarti informasi lebih banyak diteruskan.

### 2.3 Regularisasi dan pemantauan model

Implementasi ini menggunakan Dropout sebesar 0,30 pada representasi embedding dan hidden state sebelum klasifikasi untuk membantu mengurangi overfitting. Kinerja training dibandingkan dengan validation loss dan validation accuracy. Jika training terus membaik sementara validation memburuk, hal tersebut dapat menjadi indikasi overfitting; jika keduanya sama-sama buruk, model berpotensi underfitting. Batch Normalization tidak digunakan pada arsitektur ini karena eksperimen difokuskan pada embedding dan LSTM dengan dropout sebagai regularisasi.


### 2.2 Perhitungan manual LSTM (satu input dan satu output)

Perhitungan berikut adalah **contoh numerik sederhana satu time step** untuk menjelaskan alur komputasi LSTM. Bobot dan bias dipilih sebagai angka ilustrasi agar perhitungan dapat diikuti secara manual; angka ini bukan bobot hasil pelatihan model.

Diketahui:

- Input saat ini: \(x_t = 0{,}5\)
- Hidden state sebelumnya: \(h_{t-1} = 0{,}2\)
- Cell state sebelumnya: \(C_{t-1} = 0{,}1\)
- Gabungan input: \([h_{t-1},x_t]=[0{,}2,0{,}5]\)

**1. Forget gate**

Gunakan \(W_f=[0{,}5,0{,}4]\) dan \(b_f=0{,}1\).

\[
f_t=\sigma(0{,}5(0{,}2)+0{,}4(0{,}5)+0{,}1)
=\sigma(0{,}4)=0{,}5987
\]

**2. Input gate**

Gunakan \(W_i=[0{,}3,0{,}2]\) dan \(b_i=-0{,}1\).

\[
i_t=\sigma(0{,}3(0{,}2)+0{,}2(0{,}5)-0{,}1)
=\sigma(0{,}06)=0{,}5150
\]

**3. Candidate cell state**

Gunakan \(W_C=[0{,}2,0{,}6]\) dan \(b_C=0\).

\[
\widetilde{C}_t=\tanh(0{,}2(0{,}2)+0{,}6(0{,}5))
=\tanh(0{,}34)=0{,}3275
\]

**4. Memperbarui cell state**

\[
C_t=f_t C_{t-1}+i_t\widetilde{C}_t
\]

\[
C_t=(0{,}5987)(0{,}1)+(0{,}5150)(0{,}3275)=0{,}2285
\]

**5. Output gate**

Gunakan \(W_o=[0{,}4,0{,}5]\) dan \(b_o=0\).

\[
o_t=\sigma(0{,}4(0{,}2)+0{,}5(0{,}5))
=\sigma(0{,}33)=0{,}5818
\]

**6. Menghasilkan hidden state**

\[
h_t=o_t\tanh(C_t)=(0{,}5818)\tanh(0{,}2285)=0{,}1307
\]

Dengan demikian, dari satu input \(x_t=0{,}5\), contoh perhitungan menghasilkan **hidden state \(h_t\approx0{,}1307\)** dan cell state \(C_t\approx0{,}2285\). Dalam model sebenarnya, setiap gerbang memiliki banyak bobot yang dipelajari secara otomatis melalui proses training.

```python
#202432057_IvanderRisqAzmi
import os
import re
import sqlite3
import random
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score

# Menetapkan seed agar pembagian data dan inisialisasi model dapat direproduksi.
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
# Membatasi thread CPU agar eksekusi lebih stabil pada lingkungan tanpa GPU.
torch.set_num_threads(1)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print("PyTorch:", torch.__version__)
print("CUDA tersedia:", torch.cuda.is_available())
```

**Penjelasan cell 1 - Import library:** `os` dan `Path` membantu memeriksa lokasi file; `re` membersihkan teks; `sqlite3` membaca database; `Counter` menghitung frekuensi token; NumPy dan Pandas mengolah data; Matplotlib membuat grafik; PyTorch membangun dan melatih LSTM; `train_test_split` membagi data secara terstratifikasi; metrik dari scikit-learn mengukur hasil klasifikasi. Seed ditetapkan agar eksperimen lebih mudah diulang.

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>

```python
#202432057_IvanderRisqAzmi
# 1. Memuat dataset SQLite (alternatif: Tweets.csv jika file CSV digunakan).
dataset_path = Path("database.sqlite")

if dataset_path.exists():
    with sqlite3.connect(dataset_path) as connection:
        raw_df = pd.read_sql_query("SELECT * FROM Tweets", connection)
elif Path("Tweets.csv").exists():
    raw_df = pd.read_csv("Tweets.csv")
else:
    raise FileNotFoundError(
        "File database.sqlite atau Tweets.csv tidak ditemukan. "
        "Letakkan file dataset di folder yang sama dengan notebook. "
        "Di Google Colab, unggah file dataset terlebih dahulu."
    )

print("Ukuran data awal:", raw_df.shape)
print("Kolom dataset:", raw_df.columns.tolist())

# 2. Memilih kolom teks dan label, lalu membersihkan nilai yang tidak sesuai.
df = raw_df[["text", "airline_sentiment"]].copy()
df["text"] = df["text"].fillna("").astype(str)
df["airline_sentiment"] = df["airline_sentiment"].fillna("").astype(str).str.lower().str.strip()
valid_labels = ["negative", "neutral", "positive"]
df = df[df["airline_sentiment"].isin(valid_labels)].copy()
df = df[df["text"].str.strip().ne("")].reset_index(drop=True)

# Fungsi normalisasi: huruf kecil, URL dan mention diseragamkan, tanda baca dibersihkan.
def clean_tweet(text):
    text = text.lower()
    text = re.sub(r"https?://\S+|www\.\S+", " url ", text)
    text = re.sub(r"@\w+", " user ", text)
    text = text.replace("#", "")
    text = re.sub(r"[^a-z0-9'\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# Tokenisasi sederhana untuk memecah tweet menjadi urutan kata dan angka.
def tokenize(text):
    return re.findall(r"[a-z]+(?:'[a-z]+)?|[0-9]+", text)

df["clean_text"] = df["text"].apply(clean_tweet)
df["tokens"] = df["clean_text"].apply(tokenize)
df = df[df["tokens"].map(len) > 0].reset_index(drop=True)

# Label diubah menjadi bilangan: 0=Negative, 1=Neutral, 2=Positive.
label_to_id = {"negative": 0, "neutral": 1, "positive": 2}
id_to_label = {value: key.title() for key, value in label_to_id.items()}
df["label_id"] = df["airline_sentiment"].map(label_to_id).astype(int)

# 3. Membagi data menjadi train/validation/test dengan proporsi 70:10:20.
all_indices = np.arange(len(df))
all_labels = df["label_id"].to_numpy()
train_val_indices, test_indices = train_test_split(
    all_indices, test_size=0.20, random_state=SEED, stratify=all_labels
)
train_indices, val_indices = train_test_split(
    train_val_indices, test_size=0.125, random_state=SEED,
    stratify=all_labels[train_val_indices]
)

# Vocabulary hanya dibangun dari data train untuk menghindari data leakage.
PAD_TOKEN = "<PAD>"
UNK_TOKEN = "<UNK>"
PAD_IDX = 0
UNK_IDX = 1
MAX_VOCAB_SIZE = 20000
MIN_FREQUENCY = 2
MAX_LENGTH = 48

train_token_counts = Counter(
    token for index in train_indices for token in df.loc[index, "tokens"]
)
vocab = {PAD_TOKEN: PAD_IDX, UNK_TOKEN: UNK_IDX}
for token, frequency in train_token_counts.most_common():
    if frequency < MIN_FREQUENCY or len(vocab) >= MAX_VOCAB_SIZE:
        continue
    vocab[token] = len(vocab)

# Mengubah token menjadi indeks, memotong urutan yang terlalu panjang, lalu padding.
def encode_tokens(tokens):
    token_ids = [vocab.get(token, UNK_IDX) for token in tokens[:MAX_LENGTH]]
    sequence_length = max(1, len(token_ids))
    token_ids = token_ids + [PAD_IDX] * (MAX_LENGTH - len(token_ids))
    return token_ids, sequence_length

encoded = [encode_tokens(tokens) for tokens in df["tokens"]]
X_all = np.array([item[0] for item in encoded], dtype=np.int64)
lengths_all = np.array([item[1] for item in encoded], dtype=np.int64)
y_all = df["label_id"].to_numpy(dtype=np.int64)

# Membentuk TensorDataset untuk PyTorch.
def build_tensor_dataset(indices):
    return TensorDataset(
        torch.tensor(X_all[indices], dtype=torch.long),
        torch.tensor(lengths_all[indices], dtype=torch.long),
        torch.tensor(y_all[indices], dtype=torch.long)
    )

train_dataset = build_tensor_dataset(train_indices)
val_dataset = build_tensor_dataset(val_indices)
test_dataset = build_tensor_dataset(test_indices)

print("Jumlah data setelah pembersihan:", len(df))
print("Train:", len(train_dataset), "| Validation:", len(val_dataset), "| Test:", len(test_dataset))
print("Jumlah vocabulary:", len(vocab), "| Panjang maksimum urutan:", MAX_LENGTH)
print("Distribusi label:\n", df["airline_sentiment"].value_counts())
print("Contoh teks bersih:", df.loc[0, "clean_text"])

# Visualisasi distribusi label pada seluruh data yang digunakan.
label_counts = df["airline_sentiment"].value_counts().reindex(valid_labels)
plt.figure(figsize=(7, 4))
plt.bar(label_counts.index.str.title(), label_counts.values)
plt.title("Distribusi Label Sentimen pada Dataset")
plt.xlabel("Kelas Sentimen")
plt.ylabel("Jumlah Tweet")
plt.grid(axis="y", linestyle="--", alpha=0.35)
plt.show()
```

**Penjelasan cell 2 - Preprocessing:** Program membaca tabel `Tweets` dari SQLite, lalu mengambil `text` dan `airline_sentiment`. Teks kosong dan label di luar tiga kelas target dikeluarkan. Fungsi `clean_tweet` melakukan case folding, mengganti URL dan mention dengan token umum, serta menghapus tanda baca yang tidak diperlukan. Fungsi `tokenize` memecah teks menjadi token. Label diubah menjadi angka. Data dibagi secara stratified menjadi 70% training, 10% validation, dan 20% testing agar proporsi kelas tetap terjaga. Vocabulary hanya dibuat dari data training untuk mencegah kebocoran informasi dari data validation/test. Token yang tidak dikenal diberi indeks UNK, sedangkan urutan diseragamkan melalui padding hingga 64 token. Hasil akhirnya diubah menjadi tensor PyTorch.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432057_IvanderRisqAzmi
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_size, num_classes, pad_idx):
        super(SequenceModel, self).__init__()
        # Embedding memetakan setiap indeks token menjadi vektor numerik.
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=pad_idx)
        self.embedding_dropout = nn.Dropout(0.30)
        # LSTM mempelajari hubungan antartoken dalam urutan tweet.
        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            num_layers=1,
            batch_first=True
        )
        self.dropout = nn.Dropout(0.30)
        # Linear menghasilkan tiga logit untuk tiga kelas sentimen.
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x, lengths):
        embedded = self.embedding(x)
        embedded = self.embedding_dropout(embedded)
        # Packing mencegah token padding ikut diproses oleh LSTM.
        packed = nn.utils.rnn.pack_padded_sequence(
            embedded, lengths.cpu(), batch_first=True, enforce_sorted=False
        )
        _, (hidden_state, cell_state) = self.lstm(packed)
        final_hidden = hidden_state[-1]
        logits = self.fc(self.dropout(final_hidden))
        return logits

model_preview = SequenceModel(
    vocab_size=len(vocab), embedding_dim=64, hidden_size=64,
    num_classes=3, pad_idx=PAD_IDX
)
print(model_preview)
print("Jumlah parameter:", sum(parameter.numel() for parameter in model_preview.parameters()))
```

**Penjelasan cell 3 - Arsitektur model:** `nn.Embedding` mengubah token menjadi vektor berdimensi 64. Dropout 0,30 membantu mengurangi ketergantungan berlebihan pada neuron tertentu. `nn.LSTM` dengan hidden size 64 memproses urutan token. `pack_padded_sequence` membuat LSTM mengabaikan token padding berdasarkan panjang urutan sebenarnya. Hidden state terakhir diteruskan ke `nn.Linear` yang menghasilkan tiga logit kelas. Softmax tidak dipasang pada model karena `CrossEntropyLoss` sudah menggabungkan operasi log-softmax dan negative log-likelihood saat training.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432057_IvanderRisqAzmi
# 1. Hyperparameter dan DataLoader.
BATCH_SIZE = 128
EMBEDDING_DIM = 64
HIDDEN_SIZE = 64
epochs = 8
learning_rate = 0.001

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.manual_seed(SEED)  # Seed ulang agar inisialisasi model training tetap reproducible.
model = SequenceModel(
    vocab_size=len(vocab), embedding_dim=EMBEDDING_DIM,
    hidden_size=HIDDEN_SIZE, num_classes=3, pad_idx=PAD_IDX
).to(device)

# Bobot kelas dihitung dari data train untuk membantu menangani distribusi label yang tidak seimbang.
train_class_counts = torch.bincount(
    torch.tensor(y_all[train_indices]), minlength=3
).float()
class_weights = (train_class_counts.sum() / (3.0 * train_class_counts)).to(device)
criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=1e-5)

# 2. Menyimpan riwayat loss dan accuracy untuk grafik.
train_losses = []
val_losses = []
train_accuracies = []
val_accuracies = []
best_val_loss = float("inf")
best_model_state = None
patience = 3
no_improvement = 0

# 3. Training loop dan evaluasi validation pada setiap epoch.
for epoch in range(epochs):
    model.train()
    total_train_loss = 0.0
    train_correct = 0
    train_total = 0

    for batch_x, batch_lengths, batch_y in train_loader:
        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)
        optimizer.zero_grad()
        logits = model(batch_x, batch_lengths)
        loss = criterion(logits, batch_y)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        total_train_loss += loss.item() * batch_y.size(0)
        train_correct += (logits.argmax(dim=1) == batch_y).sum().item()
        train_total += batch_y.size(0)

    epoch_train_loss = total_train_loss / train_total
    epoch_train_accuracy = train_correct / train_total

    model.eval()
    total_val_loss = 0.0
    val_correct = 0
    val_total = 0
    with torch.no_grad():
        for batch_x, batch_lengths, batch_y in val_loader:
            batch_x = batch_x.to(device)
            batch_y = batch_y.to(device)
            logits = model(batch_x, batch_lengths)
            loss = criterion(logits, batch_y)
            total_val_loss += loss.item() * batch_y.size(0)
            val_correct += (logits.argmax(dim=1) == batch_y).sum().item()
            val_total += batch_y.size(0)

    epoch_val_loss = total_val_loss / val_total
    epoch_val_accuracy = val_correct / val_total
    train_losses.append(epoch_train_loss)
    val_losses.append(epoch_val_loss)
    train_accuracies.append(epoch_train_accuracy)
    val_accuracies.append(epoch_val_accuracy)

    print(
        f"Epoch {epoch + 1:02d}/{epochs} | "
        f"Train Loss: {epoch_train_loss:.4f} | Train Acc: {epoch_train_accuracy:.4f} | "
        f"Val Loss: {epoch_val_loss:.4f} | Val Acc: {epoch_val_accuracy:.4f}"
    )

    # Early stopping menyimpan bobot dengan validation loss terbaik.
    if epoch_val_loss < best_val_loss:
        best_val_loss = epoch_val_loss
        best_model_state = {key: value.detach().cpu().clone() for key, value in model.state_dict().items()}
        no_improvement = 0
    else:
        no_improvement += 1
        if no_improvement >= patience:
            print("Early stopping: validation loss tidak membaik.")
            break

if best_model_state is not None:
    model.load_state_dict(best_model_state)

print("Perangkat training:", device)
print("Epoch yang dijalankan:", len(train_losses))
print("Validation loss terbaik:", round(best_val_loss, 4))
```

**Penjelasan cell 4 - Training:** DataLoader membagi data ke dalam batch berukuran 128. Model dilatih menggunakan Adam dengan learning rate 0,001 dan CrossEntropyLoss berbobot kelas. Bobot kelas dihitung hanya dari distribusi label data training untuk mengurangi dampak ketidakseimbangan kelas. Pada setiap batch, model melakukan forward pass, menghitung loss, backpropagation, clipping gradien, dan pembaruan parameter. Setelah satu epoch, loss dan accuracy dihitung pada data training serta validation. Early stopping dengan patience 3 menghentikan training apabila validation loss tidak membaik selama tiga epoch berturut-turut, lalu bobot terbaik dipulihkan.

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432057_IvanderRisqAzmi
# 1. Grafik riwayat loss training dan validation.
epoch_axis = range(1, len(train_losses) + 1)
plt.figure(figsize=(9, 4))
plt.plot(epoch_axis, train_losses, marker="o", label="Training Loss")
plt.plot(epoch_axis, val_losses, marker="o", label="Validation Loss")
plt.title("Training dan Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend()
plt.tight_layout()
plt.savefig("training_loss.png", dpi=160, bbox_inches="tight")
plt.show()

# 2. Grafik riwayat accuracy training dan validation.
plt.figure(figsize=(9, 4))
plt.plot(epoch_axis, train_accuracies, marker="o", label="Training Accuracy")
plt.plot(epoch_axis, val_accuracies, marker="o", label="Validation Accuracy")
plt.title("Training dan Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend()
plt.tight_layout()
plt.savefig("training_accuracy.png", dpi=160, bbox_inches="tight")
plt.show()

# 3. Prediksi pada data test yang tidak digunakan selama training.
model.eval()
y_true = []
y_pred = []
y_prob = []
with torch.no_grad():
    for batch_x, batch_lengths, batch_y in test_loader:
        batch_x = batch_x.to(device)
        logits = model(batch_x, batch_lengths)
        probabilities = torch.softmax(logits, dim=1).cpu().numpy()
        predictions = probabilities.argmax(axis=1)
        y_true.extend(batch_y.numpy().tolist())
        y_pred.extend(predictions.tolist())
        y_prob.extend(probabilities.tolist())

y_true = np.array(y_true)
y_pred = np.array(y_pred)
y_prob = np.array(y_prob)
class_names = ["Negative", "Neutral", "Positive"]
test_accuracy = accuracy_score(y_true, y_pred)
test_macro_f1 = f1_score(y_true, y_pred, average="macro")
test_weighted_f1 = f1_score(y_true, y_pred, average="weighted")

print("HASIL EVALUASI DATA TEST")
print("Accuracy:", round(test_accuracy, 4))
print("Macro F1-score:", round(test_macro_f1, 4))
print("Weighted F1-score:", round(test_weighted_f1, 4))
print("\nClassification Report:")
print(classification_report(y_true, y_pred, labels=[0, 1, 2], target_names=class_names, digits=4, zero_division=0))

# 4. Visualisasi distribusi hasil prediksi dan rata-rata confidence prediksi.
predicted_counts = np.bincount(y_pred, minlength=3)
predicted_confidence = y_prob.max(axis=1)
mean_confidence_by_class = [
    float(predicted_confidence[y_pred == class_id].mean()) if np.any(y_pred == class_id) else 0.0
    for class_id in range(3)
]
fig, ax1 = plt.subplots(figsize=(9, 4.5))
bars = ax1.bar(class_names, predicted_counts, alpha=0.8, label="Jumlah Prediksi")
ax1.set_xlabel("Kelas Sentimen")
ax1.set_ylabel("Jumlah Tweet yang Diprediksi")
ax1.set_title("Distribusi Prediksi Sentimen dan Skor Keyakinan")
ax1.grid(axis="y", linestyle="--", alpha=0.35)
for bar in bars:
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height(), f"{int(bar.get_height())}", ha="center", va="bottom")
ax2 = ax1.twinx()
ax2.plot(class_names, mean_confidence_by_class, marker="o", linewidth=2, label="Rata-rata Confidence")
ax2.set_ylabel("Rata-rata Confidence")
ax2.set_ylim(0, 1)
fig.tight_layout()
plt.savefig("predicted_sentiment_distribution.png", dpi=160, bbox_inches="tight")
plt.show()

# 5. Confusion matrix untuk membandingkan label aktual dan prediksi.
cm = confusion_matrix(y_true, y_pred, labels=[0, 1, 2])
plt.figure(figsize=(6, 5))
plt.imshow(cm, interpolation="nearest")
plt.title("Confusion Matrix - Data Test")
plt.colorbar(label="Jumlah Data")
plt.xticks(range(3), class_names, rotation=20)
plt.yticks(range(3), class_names)
plt.xlabel("Prediksi")
plt.ylabel("Aktual")
for row in range(cm.shape[0]):
    for column in range(cm.shape[1]):
        plt.text(column, row, str(cm[row, column]), ha="center", va="center")
plt.tight_layout()
plt.savefig("confusion_matrix.png", dpi=160, bbox_inches="tight")
plt.show()
```
